### Handling Missing Values

Given the time-series nature of the data, forward-filling (filling with the previous valid observation) is a suitable imputation strategy for the small number of missing values in `Depth_to_Groundwater_P25` and `Volume_C10_Petrignano`.

```python
# Impute missing values using forward fill
df['Depth_to_Groundwater_P25'] = df['Depth_to_Groundwater_P25'].ffill()
df['Volume_C10_Petrignano'] = df['Volume_C10_Petrignano'].ffill()

# Recheck for missing values to confirm imputation
print("\nMissing values after imputation:")
display(df.isnull().sum())
```

### Feature Engineering: Creating Lagged Variables

For time-series forecasting, it's common to use past values of a variable as features to predict future values. Here, we create lagged features for `Depth_to_Groundwater_P25` and `Volume_C10_Petrignano`.

```python
# Sort by Date to ensure correct lagging
df = df.sort_values(by='Date').reset_index(drop=True)

# Create lagged features for Depth_to_Groundwater_P25 (target candidate)
df['Depth_to_Groundwater_P25_lag_1'] = df['Depth_to_Groundwater_P25'].shift(1)
df['Depth_to_Groundwater_P25_lag_2'] = df['Depth_to_Groundwater_P25'].shift(2)

# Create lagged features for Volume_C10_Petrignano
df['Volume_C10_Petrignano_lag_1'] = df['Volume_C10_Petrignano'].shift(1)
df['Volume_C10_Petrignano_lag_2'] = df['Volume_C10_Petrignano'].shift(2)

# Drop rows with NaN values resulting from lagging (first few rows)
df_cleaned = df.dropna().copy()

print("DataFrame with lagged features (first 5 rows after dropping NaNs):")
display(df_cleaned.head())
```

### Defining Features (X) and Target (y)

Assuming the goal is to predict `Depth_to_Groundwater_P25`, we define `y` as this column and `X` as all other relevant features, including the newly created lagged variables. The `Date` column is dropped as it's not a direct feature for most models.

```python
# Define target variable (y)
y = df_cleaned['Depth_to_Groundwater_P25']

# Define features (X)
X = df_cleaned.drop(columns=['Date', 'Depth_to_Groundwater_P25'])

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
print("\nFirst 5 rows of X:")
display(X.head())
print("\nFirst 5 rows of y:")
display(y.head())
```

### Helper Function for Cross-Validation Evaluation

To make the cross-validation evaluation more modular and reusable, I'll define a helper function `evaluate_model_selection` that takes the pipeline, data, cross-validation strategy, and a name as input, and then prints and returns the R2 scores and their mean.

```python
def evaluate_model_selection(pipe, X, y, cv_strategy, name):
    scores = cross_val_score(pipe, X, y, cv=cv_strategy, scoring='r2')
    print(f"{name} R2 Scores: {scores}")
    print(f"Mean {name} R2 Score: {scores.mean()}")
    return scores, scores.mean()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split, KFold, TimeSeriesSplit, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeRegressor

# Load the CSV file (from cell 6578799e)
file_path = '/content/Petrignano.csv'
df = pd.read_csv(file_path)

# Convert 'Date' column to datetime objects (from cell 4e9f11f3)
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Impute missing values using forward fill (from cell b634d8df)
df['Depth_to_Groundwater_P25'] = df['Depth_to_Groundwater_P25'].ffill()
df['Volume_C10_Petrignano'] = df['Volume_C10_Petrignano'].ffill()

# Sort by Date to ensure correct lagging (from cell 1e957390)
df = df.sort_values("Date").set_index("Date")
monthly = df.resample("ME").mean().reset_index()

# Create lagged features (from cell 1e957390)
df['Depth_to_Groundwater_P25_lag_1'] = df['Depth_to_Groundwater_P25'].shift(1)
df['Depth_to_Groundwater_P25_lag_2'] = df['Depth_to_Groundwater_P25'].shift(2)
df['Volume_C10_Petrignano_lag_1'] = df['Volume_C10_Petrignano'].shift(1)
df['Volume_C10_Petrignano_lag_2'] = df['Volume_C10_Petrignano'].shift(2)

# Drop rows with NaN values resulting from lagging (from cell 1e957390)
df_cleaned = df.dropna().copy()

# Define target variable (y) and features (X) (from cell 2239ec85)
y = df_cleaned['Depth_to_Groundwater_P25']
X = df_cleaned.drop(columns=['Depth_to_Groundwater_P25'])

# Split data into training and testing sets (from cell 1d46623c)
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

# Define KFold cross-validation (from cell 652bb8a5)
cv_naive = KFold(n_splits=5, shuffle=True, random_state=42)

# Define Pipeline (from cell 0b5da13f)
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('regressor', DecisionTreeRegressor(max_depth=10, random_state=42))
])

# Perform cross-validation with Naive CV (KFold) using the helper function
naive_cv_scores, mean_naive_cv_score = evaluate_model_selection(pipe, X_train_full, y_train_full, cv_naive, "Naive CV (KFold)")
```

### Hyperparameter Tuning with GridSearchCV

Now we will use `GridSearchCV` to systematically work through multiple combinations of parameter values, cross-validating each combination to determine which gives the best performance. For time series, it's crucial to use `TimeSeriesSplit` to avoid data leakage.

```python
from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    cv=cv_temporal, # Using TimeSeriesSplit for appropriate time-series cross-validation
    scoring='r2',
    n_jobs=-1 # Use all available cores
)
```
